```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from tqdm import tqdm

from MultiTaskPPGNet import MultiTaskPPGNet
from ubfc_dataset_fixed import UBFCDataset  # CHANGED: use the fixed loader
from split_utils import subject_grouped_split, save_split_subjects  # CHANGED: replaces random_split

# ==========================================
# 1. CONFIGURATION
# ==========================================
UBFC_CHUNK_DIR = "D:/MLandAi/ResearchPaper/WellnessMonitoringProject/datasets/UBFCpreProcess/UBFC_video_chunks"
STAGE1_MODEL_PATH = "models/wesad_stage1_4head.pth"
STAGE2_MODEL_PATH = "models/wesad_stage2_ada_bn.pth"
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

BATCH_SIZE = 16
EPOCHS = 20
LEARNING_RATE = 0.0005

print(f"Device: {DEVICE}")

# ==========================================
# 2. DATASET & DATALOADER
# ==========================================
dataset = UBFCDataset(UBFC_CHUNK_DIR)

# CHANGED: subject-independent split. UBFC-rPPG chunk filenames must start
# with the subject id (e.g. "subject14_...") for this to work -- verify
# against split_utils.extract_subject_id's pattern first.
train_dataset, val_dataset = subject_grouped_split(dataset, test_size=0.2, random_state=42)

# CHANGED: persist which subjects were held out, so eval_stage2_fixed.py can
# guarantee it only scores on subjects Stage 2 never trained on.
save_split_subjects(dataset, train_dataset, val_dataset, "models/ubfc_stage2_split.json")

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

print(f"Train samples: {len(train_dataset)}, Val samples: {len(val_dataset)}")

# ==========================================
# 3. LOAD STAGE 1 MODEL & APPLY AdaBN
# ==========================================
print("\n[1] Loading Stage 1 WESAD model...")
model = MultiTaskPPGNet(num_stress_classes=2).to(DEVICE)
model.load_state_dict(torch.load(STAGE1_MODEL_PATH, map_location=DEVICE, weights_only=True))
print("   Stage 1 model loaded.")

print("\n[2] Applying AdaBN (recalibrating BatchNorm statistics on UBFC data)...")
for param in model.parameters():
    param.requires_grad = False

model.train()  # BN layers update running stats in train mode
with torch.no_grad():
    for batch_x, _ in tqdm(train_loader, desc="AdaBN Pass"):
        batch_x = batch_x.to(DEVICE)
        _ = model(batch_x)

for m in model.modules():
    if isinstance(m, nn.BatchNorm1d):
        m.eval()  # freeze the newly recalibrated BN stats

print("   BatchNorm statistics recalibrated and frozen.")

# ==========================================
# 4. UNFREEZE HEADS FOR RETRAINING
# ==========================================
print("\n[3] Unfreezing HR and HRV heads for retraining...")
for param in model.hr_head.parameters():
    param.requires_grad = True
for param in model.hrv_head.parameters():
    param.requires_grad = True

optimizer = optim.Adam(
    list(model.hr_head.parameters()) + list(model.hrv_head.parameters()),
    lr=LEARNING_RATE, weight_decay=1e-5
)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=3)
criterion_mse = nn.MSELoss()

# ==========================================
# 5. TRAINING LOOP (heads only, BN frozen, dropout ACTIVE)
# ==========================================
print(f"\n[4] Starting Stage 2 Training for {EPOCHS} epochs...")
best_val_loss = float('inf')

for epoch in range(EPOCHS):
    # CHANGED: previously `model.eval()` here, which also disabled Dropout
    # in the heads being trained. Fix: put the whole model in train mode
    # (so Dropout is active for the heads), then re-freeze ONLY BatchNorm.
    model.train()
    for m in model.modules():
        if isinstance(m, nn.BatchNorm1d):
            m.eval()

    train_loss_total = 0.0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{EPOCHS}")
    for batch_x, batch_labels in pbar:
        batch_x = batch_x.to(DEVICE)
        batch_labels = {k: v.to(DEVICE) for k, v in batch_labels.items()}

        optimizer.zero_grad()
        outputs = model(batch_x)

        loss_hr = criterion_mse(outputs['hr'].squeeze(), batch_labels['hr'])
        loss_hrv = criterion_mse(outputs['hrv'].squeeze(), batch_labels['hrv'])
        loss = (0.01 * loss_hr) + (0.1 * loss_hrv)

        loss.backward()
        optimizer.step()

        train_loss_total += loss.item()
        pbar.set_postfix(loss=f"{loss.item():.4f}")

    avg_train_loss = train_loss_total / len(train_loader)

    # Validation: full eval mode is correct here (no dropout, frozen BN)
    model.eval()
    val_loss_total = 0.0
    with torch.no_grad():
        for batch_x, batch_labels in val_loader:
            batch_x = batch_x.to(DEVICE)
            batch_labels = {k: v.to(DEVICE) for k, v in batch_labels.items()}
            outputs = model(batch_x)

            loss_hr = criterion_mse(outputs['hr'].squeeze(), batch_labels['hr'])
            loss_hrv = criterion_mse(outputs['hrv'].squeeze(), batch_labels['hrv'])
            loss = (0.01 * loss_hr) + (0.1 * loss_hrv)
            val_loss_total += loss.item()

    avg_val_loss = val_loss_total / len(val_loader)
    scheduler.step(avg_val_loss)

    print(f"Epoch {epoch+1} | Train Loss: {avg_train_loss:.4f} | Val Loss: {avg_val_loss:.4f}")

    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        os.makedirs('models', exist_ok=True)
        torch.save(model.state_dict(), STAGE2_MODEL_PATH)
        print(f"  Saved best Stage 2 (AdaBN) model to {STAGE2_MODEL_PATH}")

print(f"\nStage 2 Transfer Learning Complete! Best Val Loss: {best_val_loss:.4f}")
```
