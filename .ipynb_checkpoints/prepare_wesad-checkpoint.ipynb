```python
import os
import glob
import numpy as np
import pickle
from scipy.signal import find_peaks
from tqdm import tqdm

# ==========================================
# CONFIGURATION
# ==========================================
RAW_DIR = 'D:/MLandAi/ResearchPaper/WellnessMonitoringProject/datasets/WESAD' 
OUTPUT_DIR = 'D:/MLandAi/ResearchPaper/WellnessMonitoringProject/datasets/WESADpreProcess/WESAD_chunks'
os.makedirs(OUTPUT_DIR, exist_ok=True)

CHUNK_SAMPLES = 640  # 10 seconds at 64Hz

print("="*70)
print("WESAD PREPROCESSING (Fixed 1D Array & Label Synchronization)")
print("="*70)

subject_folders = sorted(glob.glob(os.path.join(RAW_DIR, 'S*')))
print(f"Found {len(subject_folders)} subject folders.")

total_chunks = 0

for sub_folder in tqdm(subject_folders, desc="Processing Subjects"):
    sub_id = os.path.basename(sub_folder)
    pkl_files = glob.glob(os.path.join(sub_folder, '*.pkl'))
    if not pkl_files:
        continue
        
    pkl_file = pkl_files[0]

    try:
        with open(pkl_file, 'rb') as f:
            data = pickle.load(f, encoding='latin1')

        # 1. Extract BVP Signal (Wrist)
        # FIX: Use .flatten() to ensure it is strictly 1-Dimensional
        bvp_signal = np.array(data['signal']['wrist']['BVP']).flatten().astype(np.float32)
        
        # 2. Extract Labels (700 Hz)
        labels_700hz = np.array(data['label']).flatten()

        # 3. SYNCHRONIZE LABELS TO 64Hz
        # The README states labels are 700Hz, BVP is 64Hz.
        # We must map the 700Hz labels down to the 64Hz signal length.
        ratio = len(labels_700hz) / len(bvp_signal)
        label_indices = (np.arange(len(bvp_signal)) * ratio).astype(int)
        
        # Ensure we don't go out of bounds
        label_indices = np.clip(label_indices, 0, len(labels_700hz) - 1)
        synced_labels = labels_700hz[label_indices]

        # 4. Map Labels to Binary Stress (0 or 1)
        # 0=transient, 1=baseline, 2=stress, 3=amusement, 4=meditation, 5/6/7=ignore
        full_labels = np.full(len(bvp_signal), -1, dtype=np.int32)
        
        full_labels[synced_labels == 1] = 0  # Baseline -> Non-Stress
        full_labels[synced_labels == 2] = 1  # Stress -> Stress
        full_labels[synced_labels == 3] = 0  # Amusement -> Non-Stress
        full_labels[synced_labels == 4] = 0  # Meditation -> Non-Stress

        # 5. Chunk the signal
        num_chunks = len(bvp_signal) // CHUNK_SAMPLES
        
        for i in range(num_chunks):
            start_idx = i * CHUNK_SAMPLES
            end_idx = start_idx + CHUNK_SAMPLES
            
            chunk_signal = bvp_signal[start_idx:end_idx]
            chunk_labels = full_labels[start_idx:end_idx]
            
            # Require at least 80% valid labels
            valid_labels = chunk_labels[chunk_labels != -1]
            if len(valid_labels) < (CHUNK_SAMPLES * 0.8):
                continue 
            
            dominant_label = int(np.round(np.mean(valid_labels)))

            # 6. Calculate HR (chunk_signal is now guaranteed 1-D)
            peaks, _ = find_peaks(chunk_signal, distance=20) 
            
            if len(peaks) >= 2:
                rr_intervals = np.diff(peaks) / 64.0
                hr_bpm = 60.0 / np.mean(rr_intervals)
                
                if 40.0 <= hr_bpm <= 150.0:
                    chunk_name = f"{sub_id}_chunk{total_chunks:04d}"
                    
                    np.save(os.path.join(OUTPUT_DIR, f"{chunk_name}_X.npy"), chunk_signal)
                    np.save(os.path.join(OUTPUT_DIR, f"{chunk_name}_hr_y.npy"), np.array([hr_bpm], dtype=np.float32))
                    np.save(os.path.join(OUTPUT_DIR, f"{chunk_name}_stress_y.npy"), np.array([dominant_label], dtype=np.int32))
                    
                    total_chunks += 1

    except Exception as e:
        print(f"  ❌ Error processing {sub_id}: {e}")

print("\n" + "="*70)
print(f"✅ SUCCESS! Data saved to: datasets/WESADpreProcess/WESAD_chunks")
print(f"Total valid chunks created: {total_chunks}")
print("="*70)
```
