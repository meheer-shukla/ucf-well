```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from tqdm import tqdm

from MultiTaskPPGNet import MultiTaskPPGNet
from wesad_dataset import WESADDataset
from split_utils import subject_grouped_split  # CHANGED: replaces random_split

# ==========================================
# 1. CONFIGURATION
# ==========================================
CHUNK_DIR = "D:/MLandAi/ResearchPaper/WellnessMonitoringProject/datasets/WESADpreProcess/WESAD_chunks"
MODEL_SAVE_PATH = "models/wesad_stage1_4head.pth"
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

BATCH_SIZE = 64
EPOCHS = 30
LEARNING_RATE = 0.001

print(f"Device: {DEVICE}")

# ==========================================
# 2. DATASET & DATALOADER
# ==========================================
dataset = WESADDataset(CHUNK_DIR)

# CHANGED: subject-independent split instead of random_split.
# WESAD chunk filenames must start with the subject id (e.g. "S2_...")
# for this to work -- check split_utils.extract_subject_id's pattern
# against your actual filenames first.
train_dataset, val_dataset = subject_grouped_split(dataset, test_size=0.2, random_state=42)

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

print(f"Train samples: {len(train_dataset)}, Val samples: {len(val_dataset)}")

# ==========================================
# 3. MODEL & OPTIMIZER
# ==========================================
model = MultiTaskPPGNet(num_stress_classes=2).to(DEVICE)

criterion_mse = nn.MSELoss()
criterion_ce = nn.CrossEntropyLoss()

optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE, weight_decay=1e-5)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=5)

print(f"Model parameters: {sum(p.numel() for p in model.parameters()):,}")

# ==========================================
# 4. SMART MULTI-TASK LOSS FUNCTION
# ==========================================
def calculate_loss(outputs, targets):
    loss_hr = criterion_mse(outputs['hr'].squeeze(), targets['hr'])
    loss_hrv = criterion_mse(outputs['hrv'].squeeze(), targets['hrv'])
    loss_stress = criterion_ce(outputs['stress'], targets['stress'])

    valid_rr_mask = targets['rr'] != -1.0
    if valid_rr_mask.any():
        loss_rr = criterion_mse(outputs['rr'][valid_rr_mask].squeeze(), targets['rr'][valid_rr_mask])
    else:
        loss_rr = torch.tensor(0.0, device=DEVICE)

    total_loss = (0.01 * loss_hr) + (0.1 * loss_hrv) + loss_stress + loss_rr
    return total_loss, loss_hr.item(), loss_stress.item()

# ==========================================
# 5. TRAINING LOOP
# ==========================================
best_val_loss = float('inf')

for epoch in range(EPOCHS):
    model.train()
    train_loss_total = 0.0

    pbar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{EPOCHS}")
    for batch_x, batch_labels in pbar:
        batch_x = batch_x.to(DEVICE)
        batch_labels = {k: v.to(DEVICE) for k, v in batch_labels.items()}

        optimizer.zero_grad()
        outputs = model(batch_x)

        loss, hr_loss_val, stress_loss_val = calculate_loss(outputs, batch_labels)
        loss.backward()
        optimizer.step()

        train_loss_total += loss.item()
        pbar.set_postfix(loss=f"{loss.item():.4f}")

    avg_train_loss = train_loss_total / len(train_loader)

    model.eval()
    val_loss_total = 0.0
    with torch.no_grad():
        for batch_x, batch_labels in val_loader:
            batch_x = batch_x.to(DEVICE)
            batch_labels = {k: v.to(DEVICE) for k, v in batch_labels.items()}
            outputs = model(batch_x)
            loss, _, _ = calculate_loss(outputs, batch_labels)
            val_loss_total += loss.item()

    avg_val_loss = val_loss_total / len(val_loader)
    scheduler.step(avg_val_loss)

    print(f"Epoch {epoch+1} | Train Loss: {avg_train_loss:.4f} | Val Loss: {avg_val_loss:.4f}")

    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        os.makedirs('models', exist_ok=True)
        torch.save(model.state_dict(), MODEL_SAVE_PATH)
        print(f"  Saved best Stage 1 model to {MODEL_SAVE_PATH}")

print(f"\nStage 1 Pretraining Complete! Best Val Loss: {best_val_loss:.4f}")
```
