```python
import numpy as np
import torch
from torch.utils.data import DataLoader
from scipy import stats
from tqdm import tqdm

from MultiTaskPPGNet import MultiTaskPPGNet
from ubfc_dataset import UBFCDataset  # CHANGED: use the fixed loader
from split_utils import load_val_subject_filter  # CHANGED

# ==========================================
# 1. CONFIGURATION
# ==========================================
UBFC_CHUNK_DIR = "D:/MLandAi/ResearchPaper/WellnessMonitoringProject/datasets/UBFCpreProcess/UBFC_video_chunks"
SPLIT_PATH = "models/ubfc_stage2_split.json"
STAGE1_MODEL_PATH = "models/wesad_stage1_4head.pth"
STAGE2_MODEL_PATH = "models/wesad_stage2_ada_bn.pth"
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

print("=" * 70)
print("EVALUATING STAGE 1 vs STAGE 2 ON HELD-OUT UBFC-rPPG SUBJECTS")
print("=" * 70)

# ==========================================
# 2. LOAD DATA -- HELD-OUT SUBJECTS ONLY
# ==========================================
full_dataset = UBFCDataset(UBFC_CHUNK_DIR)

# CHANGED: previously this evaluated on `full_dataset` directly, which
# included the ~80% of chunks Stage 2 was fine-tuned on. Now we only score
# on subjects saved as "val" during Stage 2 training -- true held-out data
# for BOTH models, making this a fair, leak-free comparison.
held_out_dataset = load_val_subject_filter(full_dataset, SPLIT_PATH)
print(f"Evaluating on {len(held_out_dataset)} held-out chunks "
      f"(subjects never used in Stage 2 fine-tuning).")

eval_loader = DataLoader(held_out_dataset, batch_size=16, shuffle=False, num_workers=0)

# ==========================================
# 3. HELPER FUNCTION TO EVALUATE A MODEL
# ==========================================
def evaluate_model(model_path, model_name):
    print(f"\n[Testing {model_name}...]")

    model = MultiTaskPPGNet(num_stress_classes=2).to(DEVICE)
    model.load_state_dict(torch.load(model_path, map_location=DEVICE, weights_only=True))
    model.eval()

    all_true_hr, all_pred_hr = [], []
    with torch.no_grad():
        for batch_x, batch_labels in tqdm(eval_loader, desc=f"Evaluating {model_name}"):
            batch_x = batch_x.to(DEVICE)
            outputs = model(batch_x)

            pred_hr = np.atleast_1d(outputs['hr'].cpu().squeeze().numpy())
            true_hr = np.atleast_1d(batch_labels['hr'].numpy())

            all_pred_hr.extend(pred_hr)
            all_true_hr.extend(true_hr)

    all_true_hr = np.array(all_true_hr)
    all_pred_hr = np.array(all_pred_hr)
    abs_errors = np.abs(all_true_hr - all_pred_hr)
    mae = abs_errors.mean()

    # CHANGED: bootstrap 95% CI -- a single MAE number with no interval
    # can't tell you if a 0.31 BPM "improvement" is real or noise.
    rng = np.random.default_rng(42)
    boot_means = [rng.choice(abs_errors, size=len(abs_errors), replace=True).mean()
                  for _ in range(2000)]
    ci_low, ci_high = np.percentile(boot_means, [2.5, 97.5])

    print(f"   {model_name} MAE: {mae:.2f} BPM  (95% CI [{ci_low:.2f}, {ci_high:.2f}])")
    return mae, abs_errors

# ==========================================
# 4. RUN EVALUATION
# ==========================================
mae_stage1, err_stage1 = evaluate_model(STAGE1_MODEL_PATH, "Stage 1 (WESAD Pretrained)")
mae_stage2, err_stage2 = evaluate_model(STAGE2_MODEL_PATH, "Stage 2 (AdaBN Transfer)")

# ==========================================
# 5. PAIRED SIGNIFICANCE TEST
# ==========================================
# CHANGED: valid now that both models are scored on the identical held-out
# chunks -- Wilcoxon signed-rank test on the paired per-chunk errors.
if len(err_stage1) == len(err_stage2):
    stat, p_value = stats.wilcoxon(err_stage1, err_stage2)
else:
    p_value = None

# ==========================================
# 6. FINAL RESULTS
# ==========================================
print("\n" + "=" * 70)
print("FINAL RESULTS FOR PAPER (held-out subjects only)")
print("=" * 70)
print(f"Baseline (Stage 1) MAE: {mae_stage1:.2f} BPM")
print(f"After Transfer (Stage 2) MAE: {mae_stage2:.2f} BPM")
print(f"Improvement: {mae_stage1 - mae_stage2:.2f} BPM")
if p_value is not None:
    print(f"Wilcoxon signed-rank p-value: {p_value:.4f}")
    if p_value >= 0.05:
        print("NOT statistically significant at alpha=0.05 -- "
              "do not report this improvement as a real effect yet.")
print("=" * 70)
```
